1. data = {}

YOLOv8 학습과 검증에 사용되는 데이터 디렉터리 지정.

2. yaml 파일 작성

yaml.dump() : data 딕셔너리를 YAML 형식으로 직렬화하여 파일에 기록함.

3. yaml 파일 열기

yaml.safe_load() : yaml 데이터를 파이썬 객체로 변환하여 로드함.

```python
import yaml

data = { 'train' : '/content/PKLot.v1-raw.yolov8/train',
        'val' : '/content/PKLot.v1-raw.yolov8/valid',
         'test' : '/content/PKLot.v1-raw.yolov8/test',
         'names' : ['space-empty', 'space-occupied'],
         'nc' : 2 }

with open('/content/PKLot.v1-raw.yolov8/PKLot.v1-raw.yolov8.yaml', 'w') as f:
  yaml.dump(data, f)

with open('/content/PKLot.v1-raw.yolov8/PKLot.v1-raw.yolov8.yaml', 'r') as f:
  PKLot_yaml = yaml.safe_load(f)
  display(PKLot_yaml)
```

```python
# Pre-trained model 로드하기
from ultralytics import YOLO
model = YOLO('yolov8n.pt')
```

```python
# 학습시키기
model.train(data='/content/PKLot.v1-raw.yolov8/PKLot.v1-raw.yolov8.yaml', epochs=10, patience=30, batch=32, imgsz=416)
```
